First follow the Gurobi documentation to install and register the notebook kernel.
Open this notebook in JupyterHub, select **Python (gurobi)**, and run the cell below.

Expected result: `x=2.0, y=2.0, objective=10.0`.
The example uses one solver thread and closes its Gurobi environment when finished.
The WLS token can still count toward the session limit for up to five more minutes with default settings.

When finished, save the notebook and shut down its kernel from JupyterLab's **Running** panel.


```python
import gurobipy as gp

with gp.Env() as env, gp.Model("example", env=env) as model:
    model.Params.Threads = 1
    x = model.addVar(ub=2, name="x")
    y = model.addVar(name="y")
    model.addConstr(x + y <= 4)
    model.setObjective(3*x + 2*y, gp.GRB.MAXIMIZE)
    model.optimize()

    print(f"x={x.X}, y={y.X}, objective={model.ObjVal}")
```
